# RideShare — Python DB-API Analytics
Synthetic Bengaluru ride-hailing dataset. Install `mysql-connector-python pandas matplotlib`. Set MYSQL_HOST, MYSQL_USER, MYSQL_PASSWORD and MYSQL_DATABASE.

```python
import os, mysql.connector, pandas as pd, matplotlib.pyplot as plt
conn=mysql.connector.connect(host=os.getenv('MYSQL_HOST','localhost'),user=os.getenv('MYSQL_USER','root'),password=os.getenv('MYSQL_PASSWORD',''),database=os.getenv('MYSQL_DATABASE','rideshare'))
cur=conn.cursor()
print('Connected to RideShare')
```

```python
sql="SELECT DATE_FORMAT(requested_at,'%Y-%m') month, COUNT(*) trips FROM trips GROUP BY month ORDER BY month"
cur.execute(sql); df=pd.DataFrame(cur.fetchall(),columns=[c[0] for c in cur.description]); df.head()
```

```python
plt.figure(figsize=(9,4)); plt.plot(df.month,df.trips,marker='o'); plt.title('Monthly Trip Requests'); plt.xticks(rotation=45); plt.ylabel('Trips'); plt.tight_layout(); plt.show()
```

```python
sql="SELECT z.zone_name, COUNT(*) trips FROM trips t JOIN zones z ON t.pickup_zone_id=z.zone_id GROUP BY z.zone_id,z.zone_name ORDER BY trips DESC"
cur.execute(sql); zones_df=pd.DataFrame(cur.fetchall(),columns=[c[0] for c in cur.description]); zones_df
```

```python
plt.figure(figsize=(8,4)); plt.bar(zones_df.zone_name,zones_df.trips); plt.title('Trips by Pickup Zone'); plt.xticks(rotation=35); plt.tight_layout(); plt.show()
```

```python
cur.close(); conn.close(); print('Connection closed')
```
